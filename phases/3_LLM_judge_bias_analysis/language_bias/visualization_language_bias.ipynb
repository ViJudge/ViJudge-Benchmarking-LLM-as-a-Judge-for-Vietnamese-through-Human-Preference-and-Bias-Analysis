```python
path = 'drive/MyDrive/ViJudge/data/language_bias/statistics/'
output_path = 'drive/MyDrive/ViJudge/data/visualization/language_bias/'
```

```python
kappa_df = pd.read_csv(path + 'kappa_result.csv')
kappa_df.head()
```

## Cohen's kappa

```python
# Calculate mean values
avg_kappa_df = kappa_df.groupby(['judge'])[['VN-VN vs EN-EN', 'VN-VN vs VN2-VN2']].mean().reset_index()

# Calculate standard deviation values
std_kappa_df = kappa_df.groupby(['judge'])[['VN-VN vs EN-EN', 'VN-VN vs VN2-VN2']].std().reset_index()

# Prepare DataFrame for display and export with formatted mean and std
display_df = avg_kappa_df.copy()

# Merge standard deviations, ensuring column names are distinct
display_df = pd.merge(display_df, std_kappa_df, on=['judge'], suffixes=('_mean', '_std'))

# Format the columns into "Avg% (±Std%)"
display_df['VN-VN vs EN-EN'] = display_df.apply(lambda row: f"{row['VN-VN vs EN-EN_mean']:.4f} (±{row['VN-VN vs EN-EN_std']:.4f})", axis=1)
display_df['VN-VN vs VN2-VN2'] = display_df.apply(lambda row: f"{row['VN-VN vs VN2-VN2_mean']:.4f} (±{row['VN-VN vs VN2-VN2_std']:.4f})", axis=1)

# Select and reorder columns for final display/export
display_df = display_df[['judge', 'VN-VN vs EN-EN', 'VN-VN vs VN2-VN2']]

# Save the formatted display DataFrame to CSV
display_df.to_csv(output_path + 'language_avg_kappa_ratio_table.csv', index=False)

# Display the formatted DataFrame
display(display_df)
```
