```python
import pandas as pd
```

```python
path = 'drive/MyDrive/ViJudge/data/verbosity_bias/statistics/'
output_path = 'drive/MyDrive/ViJudge/data/visualization/verbosity_bias/'
```

```python
bias_df = pd.read_csv(path + 'bias_by_attempt_and_config_result.csv')
bias_df.head()
```

## Avg bias ratio

```python
# Calculate mean values
avg_bias_df = bias_df.groupby(['judge', 'config'])[['not_bias', 'bias', 'error']].mean().reset_index()

# Calculate standard deviation values
std_bias_df = bias_df.groupby(['judge', 'config'])[['not_bias', 'bias', 'error']].std().reset_index()

# Prepare DataFrame for display and export with formatted mean and std
display_df = avg_bias_df.copy()

# Merge standard deviations, ensuring column names are distinct
display_df = pd.merge(display_df, std_bias_df, on=['judge', 'config'], suffixes=('_mean', '_std'))

# Format the columns into "Avg% (±Std%)"
display_df['Not bias (%)'] = display_df.apply(lambda row: f"{row['not_bias_mean']*100:.2f} (±{row['not_bias_std']*100:.2f})", axis=1)
display_df['Bias (%)'] = display_df.apply(lambda row: f"{row['bias_mean']*100:.2f} (±{row['bias_std']*100:.2f})", axis=1)
display_df['Error (%)'] = display_df.apply(lambda row: f"{row['error_mean']*100:.2f} (±{row['error_std']*100:.2f})", axis=1)

# Select and reorder columns for final display/export
display_df = display_df[['judge', 'config', 'Not bias (%)', 'Bias (%)', 'Error (%)']]

# Save the formatted display DataFrame to CSV
display_df.to_csv(output_path + 'verbosity_avg_bias_ratio_table.csv', index=False)

# Display the formatted DataFrame
display(display_df)
```
