```python
import pandas as pd
import ast
```

```python
path = 'drive/MyDrive/ViJudge/data/data_for_agreement/'
output_path = 'drive/MyDrive/ViJudge/data/visualization/agreement/'
```

```python
results_df = pd.read_csv(path + 'results.csv')
results_df.head()
```

## Ties

```python
def parse_ties(row):
    ties_dict = ast.literal_eval(row['Ties'])
    return pd.Series([
        ties_dict['AI_vs_Human'],
        ties_dict['Human_vs_Human'],
        ties_dict['Upper_limit']
    ])
```

```python
ties_df = results_df.copy()
ties_df[['AI_vs_Human', 'Human_vs_Human', 'Upper_limit']] = results_df.apply(parse_ties, axis=1)
ties_df.drop(columns=['Ties', 'No_ties', 'Category'], inplace=True)
ties_df.to_csv(output_path + 'agreement_ties.csv', index=False)
ties_df
```

## No ties

```python
def parse_no_ties(row):
    no_ties_dict = ast.literal_eval(row['No_ties'])
    return pd.Series([
        no_ties_dict['AI_vs_Human'],
        no_ties_dict['Human_vs_Human'],
        no_ties_dict['Upper_limit']
    ])
```

```python
no_ties_df = results_df.copy()
no_ties_df[['AI_vs_Human', 'Human_vs_Human', 'Upper_limit']] = results_df.apply(parse_no_ties, axis=1)
no_ties_df.drop(columns=['Ties', 'No_ties', 'Category'], inplace=True)
no_ties_df.to_csv(output_path + 'agreement_no_ties.csv', index=False)
no_ties_df
```

## Category

```python
def parse_category_data(row):
    category_dict = ast.literal_eval(row['Category'])
    parsed_data = {}
    for category, (agreed, total, ratio) in category_dict.items():
        # Format as "ratio (agreed/total)" rounded to 3 decimal places
        parsed_data[category] = f"{ratio:.3f} ({agreed}/{total})"
    return pd.Series(parsed_data)
```

```python
category_df = results_df.copy()

# Apply the parsing function to create new columns for each category
category_data_cols = category_df.apply(parse_category_data, axis=1)

# Concatenate the new category columns with 'model' and 'type'
category_df = pd.concat([category_df[['model', 'type']], category_data_cols], axis=1)

category_df.to_csv(output_path + 'agreement_category.csv', index=False)
category_df
```
